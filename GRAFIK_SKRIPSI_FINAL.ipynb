# Grafik Skripsi Model Final

Notebook ini merangkum grafik-grafik utama untuk skripsi Web-GIS prediksi curah hujan 4 class BMKG.

- Model operasional final: `2005+ / time_steps=3 / focal`
- Accuracy final: `0.6304`
- Macro F1 final: `0.3708`
- Recall kelas 3 final: `0.1429`
- Folder grafik: `artifacts/skripsi_graphs_20260706_122221`


## Ringkasan Model Utama

- Baseline lama: accuracy `0.5834`, macro F1 `0.3476`, recall kelas 3 `0.1304`.
- Final 2005+: accuracy `0.6304`, macro F1 `0.3708`, recall kelas 3 `0.1429`.
- 2010+ terbaik: accuracy `0.5764`, macro F1 `0.3475`, recall kelas 3 `0.0500`.
